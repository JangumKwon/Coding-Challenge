```python
#DAY2. 문자열 압축하기(Python) [수정버전-1]
# 문제설명: 문자열이 주어졌을 때, 연속해서 등장하는 동일한 문자를 하나의 문자와 반복 횟수로 압축하는 프로그램을 작성하시오.
# 제한조건: 1=<문자열 길이=<1,000 / 문자열은 영문 소문자(a~z), 동일한 문자가 연속해서 등장하는 경우만 압축, 압축 결과가 원본 문자열보다 길어져도 그대로 출력
# 출력조건: 주어진 문자열을 압축한 결과를 한 줄에 출력한다.

x = input()
N = len(x)

Count = 1      #Count는 1로 초기화
for i in range(N-1):   # range로 범위 지정 후 반복
    if x[i] == x[i+1]:
         Count += 1  # += 순서 주의!
         result = x[i] + str(Count)   # Count 문자열로 변경 필요
    else:
         Count = 1
         result = result + x[i+1]

print(result)
```

```python
#DAY2. 문자열 압축하기(Python) [수정버전-3, 정답처리]
# 문제설명: 문자열이 주어졌을 때, 연속해서 등장하는 동일한 문자를 하나의 문자와 반복 횟수로 압축하는 프로그램을 작성하시오.
# 제한조건: 1=<문자열 길이=<1,000 / 문자열은 영문 소문자(a~z), 동일한 문자가 연속해서 등장하는 경우만 압축, 압축 결과가 원본 문자열보다 길어져도 그대로 출력
# 출력조건: 주어진 문자열을 압축한 결과를 한 줄에 출력한다.

x = input()
N = len(x)

count = 1      #Count는 1로 초기화
result = ""    #result 초기화
for i in range(N-1):   # range로 범위 지정 후 반복
    if x[i] == x[i+1]:
         count += 1  # += 순서 주의!

    else:
        if count > 1:
            result += x[i] + str(count) # Count 문자열로 변경 필요 및 = 와 += 차이 주의
        else:
            result += x[i]

        count = 1

last_result = ""
if count > 1:
    last_result += x[-1] + str(count)
else:
    last_result += x[-1]

print(result + last_result)
```

```python
# DAY 2. 문자열 압축하기 (Python) - 개선 버전

# 1. 문자열 입력 및 변수 초기화
x = input()
N = len(x)

count = 1
result = ""

# 2. 인접 문자 비교 (반복문)
for i in range(N - 1):

    # 2-1. 동일한 문자일 경우: 반복 횟수 증가
    if x[i] == x[i + 1]:
        count += 1

     # 2-2. 다른 문자일 경우: 이전 문자 그룹 저장
    else:
        result += x[i]

        if count > 1:
            result += str(count)

        count = 1  # 새로운 문자 그룹을 위해 초기화

# 3. 마지막 문자 그룹 처리
result += x[-1]

if count > 1:
    result += str(count)

# 4. 최종 결과 출력
print(result)
```

```python
# DAY 2. 문자열 압축하기 (Python) - 간결 버전

x = input()
result = ""
count = 1

for i in range(1, len(x)):
    if x[i] == x[i - 1]:
        count += 1
    else:
        result += x[i - 1] + (str(count) if count > 1 else "")
        count = 1

result += x[-1] + (str(count) if count > 1 else "")

print(result)
```
